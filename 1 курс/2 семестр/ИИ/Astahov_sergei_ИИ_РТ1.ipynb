# 4.3.1

```python
import numpy as np
import matplotlib.pyplot as plt

aray_size = 100
random_aray = np.random.rand(aray_size)
mean_values = np.mean(random_aray)
median_values = np.median(random_aray)

print(f'Ср. значение массива: {mean_values}')
print(f'Медиана массива: {median_values}')
plt.scatter(range(len(random_aray)), random_aray, color='g', label='Случайные значения')
plt.axhline(y = mean_values, color='r', linestyle='-', label = 'Ср.знач')
plt.axhline(y = median_values, color = 'b', linestyle='--', label = 'Медиана')

plt.xlabel('Индекс')
plt.ylabel('Значение')
plt.legend()
plt.show()
```

# 4.3.5

```python
import math

def calculator(x, y, operator):
    result = None
    if operator == '+':
        result = x + y
    elif operator == '-':
        result = x - y
    elif operator == '*':
        result = x * y
    elif operator == '/':
        if y != 0:
            result = x / y
        else:
            result = "Деление на ноль!"
    elif operator == 'exp':
        result = math.exp(x + y)
    elif operator == 'sin':
        result = math.sin(x + y)
    elif operator == 'cos':
        result = math.cos(x + y)
    elif operator == 'xy':
        result = x ** y
    else:
        result = "Неподдерживаемая операция"
    return result

# Пример использования калькулятора
x = 12
y = 9

print("Сложение:", calculator(x, y, '+'))
print("Вычитание:", calculator(x, y, '-'))
print("Умножение:", calculator(x, y, '*'))
print("Деление:", calculator(x, y, '/'))
print("Экспонента:", calculator(x, y, 'exp'))
print("Синус суммы:", calculator(x, y, 'sin'))
print("Косинус суммы:", calculator(x, y, 'cos'))
print("Возведение в степень:", calculator(x, y, 'xy'))
```
